````python
# 1. Initialize Inventory with Error Handling
def initialize_inventory(products):
    inventory = {}
    for product in products:
        valid_quantity = False
        while not valid_quantity:
            try:
                quantity = int(input(f"Enter the quantity of {product}s available: "))
                if quantity < 0:
                    print("Error: Quantity cannot be negative. Please enter a valid quantity.")
                else:
                    inventory[product] = quantity
                    valid_quantity = True
            except ValueError:
                print("Error: Invalid input. Please enter a valid integer quantity.")
    return inventory


# 2. Get Customer Orders with Error Handling (Note the 'inventory' parameter!)
def get_customer_orders(inventory):
    customer_orders = set()
    
    # Validate number of orders
    valid_num = False
    while not valid_num:
        try:
            num_orders = int(input("Enter the number of customer orders: "))
            if num_orders <= 0:
                print("Please enter a number greater than 0.")
            else:
                valid_num = True
        except ValueError:
            print("Invalid input. Please enter a valid number.")
            
    # Validate product selection
    for i in range(num_orders):
        valid_product = False
        while not valid_product:
            product = input("Enter the name of a product that a customer wants to order: ").strip().lower()
            
            if product not in inventory:
                print(f"Error: '{product}' is not available in our store inventory. Please choose a valid item.")
            elif inventory[product] <= 0:
                print(f"Error: '{product}' is currently out of stock. Please select a different product.")
            else:
                customer_orders.add(product)
                valid_product = True
                
    return customer_orders


# 3. Calculate Order Statistics
def calculate_order_statistics(customer_orders, products):
    total_ordered = len(customer_orders)
    percentage = (total_ordered / len(products)) * 100
    return total_ordered, percentage


# 4. Print Order Statistics
def print_order_statistics(order_statistics):
    print("\nOrder Statistics:")
    print(f"Total Products Ordered: {order_statistics[0]}")
    print(f"Percentage of Unique Products Ordered: {order_statistics[1]}")


# 5. Update Inventory (filtering out stock <= 0)
def update_inventory(customer_orders, inventory):
    for product in customer_orders:
        if product in inventory:
            inventory[product] -= 1
    return {product: qty for product, qty in inventory.items() if qty > 0}


# 6. Print Updated Inventory
def print_updated_inventory(inventory):
    print("\nUpdated Inventory:")
    for product, quantity in inventory.items():
        print(f"{product}: {quantity}")


# 7. Calculate Total Price with Error Handling
def calculate_total_price(customer_orders):
    total_price = 0.0
    for product in customer_orders:
        valid_price = False
        while not valid_price:
            try:
                price = float(input(f"Enter the price of {product}: "))
                if price < 0:
                    print("Price cannot be negative. Please enter a valid price.")
                else:
                    total_price += price
                    valid_price = True
            except ValueError:
                print("Invalid input. Please enter a numerical value for price.")
                
    print(f"Total Price: {total_price}")
    return total_price
````

````python
products = ["t-shirt", "mug", "hat", "book", "keychain"]

# Step 1: Initialize Inventory
inventory = initialize_inventory(products)

# Step 2: Get Customer Orders
customer_orders = get_customer_orders(inventory)

# Step 3 & 4: Order Statistics
stats = calculate_order_statistics(customer_orders, products)
print_order_statistics(stats)

# Step 5 & 6: Update & Print Inventory
inventory = update_inventory(customer_orders, inventory)
print_updated_inventory(inventory)

# Step 7: Calculate Total Price
total_price = calculate_total_price(customer_orders)
````
